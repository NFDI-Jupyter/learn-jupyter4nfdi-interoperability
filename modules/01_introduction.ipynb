##  Problems with interoperability

Let's assume that temperature observations are published by two independent organisations. The datasets look simple and both are CSV files containing dates, locations and temperatures. However, can we combine them to get more data?



```python
import pandas as pd

a = pd.read_csv("../data/dataset_a.csv")
b = pd.read_csv("../data/dataset_b.csv")

print("Dataset A")
display(a.head())

print("Dataset B")
display(b.head())
```

Let's look at the two tables.

- Does `station` mean the same thing as `site_id`?
- Does `temp` mean the same thing as `temperature`?
- Are the temperatures measured in the same unit?
- Could we safely combine the files without extra information?

```python
print(a.columns.tolist())
print(b.columns.tolist())
```

Dataset A stores temperature in **degrees Celsius**.

Dataset B stores temperature in **degrees Fahrenheit**.

The numbers looked plausible in both files, so without documentation we could easily make a mistake.

```python
# Rename the obvious matching columns in Dataset A
a2 = a.rename(columns={
    "station": "station_id",
    "temp": "temperature",
    "date": "observation_date"
})

# Convert Dataset B temperature from Fahrenheit to Celsius
b2 = b.copy()
b2["temperature"] = (b2["temperature"] - 32) * 5/9

display(a2.head())
display(b2.head())
```

Two CSV files are not automatically interoperable.

To reuse data safely, we often need to know:

- what each variable means;
- which units are used;
- what identifiers refer to;
- what coded values mean;
- how the data were created.

These are exactly the kinds of problems that interoperability frameworks such as **CDIF** try to address.
